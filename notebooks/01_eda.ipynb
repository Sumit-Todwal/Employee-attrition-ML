```python
def attrition_rate_by_column(df, column):
    return (
        pd.crosstab(
            df[column],
            df["Attrition"],
            normalize="index"
        ) * 100
    ).round(2)
```

```python
def attrition_summary(df, column):
    summary = pd.crosstab(
        df[column],
        df["Attrition"]
    )

    summary["Total"] = summary.sum(axis=1)
    summary["AttritionRate"] = (
        summary["Yes"] / summary["Total"] * 100
    ).round(2)

    return summary.sort_values("AttritionRate", ascending=False)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
```

```python
numeric_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)
```

```python
categorical_transformer = Pipeline(
    steps=[
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            )
        )
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                random_state=42
            )
        )
    ]
)
```

```python
logistic_pipeline.fit(X_train, y_train)
```

```python
y_pred = logistic_pipeline.predict(X_test)
```

```python
y_proba = logistic_pipeline.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.20, 0.30, 0.40, 0.50, 0.60, 0.70]

results = []

for threshold in thresholds:
    y_pred_threshold = np.where(
        y_proba >= threshold,
        "Yes",
        "No"
    )

    results.append({
        "Threshold": threshold,
        "Precision": precision_score(
            y_test,
            y_pred_threshold,
            pos_label="Yes"
        ),
        "Recall": recall_score(
            y_test,
            y_pred_threshold,
            pos_label="Yes"
        ),
        "F1": f1_score(
            y_test,
            y_pred_threshold,
            pos_label="Yes"
        )
    })

threshold_results = pd.DataFrame(results)

threshold_results
```

```python
def subgroup_metrics(df, group_column):

    results = []

    for group, group_df in df.groupby(group_column):

        actual = group_df["Actual"]
        predicted = group_df["Predicted"]

        actual_leavers = (actual == "Yes").sum()

        # Recall is meaningful only when the group contains
        # at least one actual "Yes"
        if actual_leavers > 0:

            recall = recall_score(
                actual,
                predicted,
                pos_label="Yes"
            )

        else:
            recall = np.nan

        predicted_leavers = (predicted == "Yes").sum()

        if predicted_leavers > 0:

            precision = precision_score(
                actual,
                predicted,
                pos_label="Yes"
            )

        else:
            precision = np.nan

        results.append({
            "Group": group,
            "Employees": len(group_df),
            "Actual_Leavers": actual_leavers,
            "Predicted_Leavers": predicted_leavers,
            "Recall": recall,
            "Precision": precision
        })

    return pd.DataFrame(results)
```

```python
jobrole_results = subgroup_metrics(
    validation_df,
    "JobRole"
)

jobrole_results.sort_values(
    "Recall",
    ascending=False
)
```
